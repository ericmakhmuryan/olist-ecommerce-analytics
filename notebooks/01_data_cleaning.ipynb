Customers Dataframe cleaning

```python
customers.info()
```

Trimming whitespace and standardizing casing

```python
customers['customer_city'] = customers['customer_city'].str.strip().str.title()
customers['customer_state'] = customers['customer_state'].str.strip().str.upper()
```

Keeping essential location columns for joins and Tableau geo-mapping

```python
customers_clean = customers[['customer_id', 'customer_unique_id', 'customer_city', 'customer_state']].copy()
```

Orders dataframe cleaning

```python
orders.info()
```

We should convert all timestamp columns from string to datetime

```python
date_columns = [
    'order_purchase_timestamp', 
    'order_approved_at', 
    'order_delivered_carrier_date', 
    'order_delivered_customer_date', 
    'order_estimated_delivery_date'
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
    
```

Checking for null values

```python
orders.isna().sum()
```

Missing values are primarily located in the columns corresponding to the order approval time, handover to the carrier, and final delivery to the customer. Consequently, these are likely not true missing values (data errors), but rather indicate orders that were canceled before reaching these fulfillment stages.

Filtering for completed transactions only


Excludes canceled, unavailable, created, or processing orders

```python
orders_clean = orders[orders['order_status'] == 'delivered'].copy()
```

Dropping rows where critical delivery timestamps are missing

```python
orders_clean = orders_clean.dropna(subset=['order_delivered_customer_date', 'order_approved_at'])
```

Actual delivery time in fractional days

```python
orders_clean['delivery_time_days'] = (
    orders_clean['order_delivered_customer_date'] - orders_clean['order_purchase_timestamp']
).dt.total_seconds() / (24 * 3600)
```

Delivery delay / early arrival (Positive = Early, Negative = Delayed)

```python
orders_clean['estimated_vs_actual_days'] = (
    orders_clean['order_estimated_delivery_date'] - orders_clean['order_delivered_customer_date']
).dt.total_seconds() / (24 * 3600)
```

Payments dataframe cleaning

```python
payments.info()
```

Checking for null values

```python
payments.isna().sum()
```

This means that we don't have null values

Preventing cartesian explosion during joins by aggregating multiple payments per order

```python
payments_clean = payments.groupby('order_id').agg(
    total_payment_value=('payment_value', 'sum'),
    max_payment_installments=('payment_installments', 'max'),
    primary_payment_type=('payment_type', 'first'),
    payment_methods_count=('payment_sequential', 'count')
).reset_index()

```

Products dataframe cleaning

```python
products_clean = products.copy()
```

Imputing missing product categories

```python
products_clean['product_category_name'] = products_clean['product_category_name'].fillna('unknown')
```

Imputing missing product numerical attributes by median value


```python
dimension_cols = ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
for col in dimension_cols:
    products_clean[col] = products_clean[col].fillna(products_clean[col].median())
```

Items dataframe cleaning

```python
items_clean = items.copy()
```

Converting shipping limit date to datetime

```python
items_clean['shipping_limit_date'] = pd.to_datetime(items_clean['shipping_limit_date'])
```

Ensuring that all financial metrics are positive numbers

```python
items_clean = items_clean[(items_clean['price'] > 0) & (items_clean['freight_value'] >= 0)]
```

Merging into the Master Dataset

```python
master_df = orders_clean.merge(customers_clean, on='customer_id', how='inner')
master_df = master_df.merge(items_clean, on='order_id', how='inner')
master_df = master_df.merge(products_clean, on='product_id', how='left')
master_df = master_df.merge(payments_clean, on='order_id', how='left')
```

Fixing Logistics Anomalies (Removing negative delivery days and extreme delays > 90 days)

```python
master_df = master_df[
    (master_df['delivery_time_days'] >= 0) & 
    (master_df['delivery_time_days'] <= 90)
]
```

Saving Master CSV for PostgreSQL & Tableau

```python
master_df.to_csv('../data/olist_master_clean.csv', index=False)
```
